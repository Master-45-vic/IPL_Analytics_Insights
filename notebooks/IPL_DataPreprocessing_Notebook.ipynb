## 3. Data Preprocessing

3.1 Date Preprocessing

* Convert Date from Object Type to Date Time format

* Sort the data according to the date

```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df = df.sort_values(by='date').reset_index(drop=True)
display(df.head())
```

3.2 Season Column Preprocessing

```python
df['season'].unique()
```

* Data Type Mismatch -> The same season appears as both an integer and a string (e.g., '2001' as string and 2001 as int)


* Convert all to string data type.

```python
df['season'] = df['season'].astype(str)
df['season'].unique()
```

3.3 Event Preprocessing

* Since all data belongs to the same event (i.e., Indian Premier League)

* There is no use of having it. We will remove it.

```python
df = df.drop(columns=['event'])
print(df.columns)
```

* Franchise Name Inconsistency

3.8 Player Name Preprocessing

```python
columns_to_check = ['player_of_match', 'batter', 'bowler', 'non_striker', 'wicket_kind', 'wicket_player_out']

for col in columns_to_check:
    print(f"\nUnique values in '{col}':")
    print(df[col].unique())
```

```python
columns_to_check_nulls = ['player_of_match', 'batter', 'bowler', 'non_striker', 'wicket_kind', 'wicket_player_out']

for col in columns_to_check_nulls:
    print(f"\nNumber of null values in '{col}': {df[col].isnull().sum()}")
```

```python
df['player_of_match']=df['player_of_match'].fillna("No Award", inplace=True)
df['wicket_kind'] = df['wicket_kind'].fillna('No Wicket')
df['wicket_player_out'] = df['wicket_player_out'].fillna('No Dismissal')
```
